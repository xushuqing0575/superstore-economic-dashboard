# Superstore · Economic Performance

R · Shiny · tidyverse · Plotly

Combined Superstore sales with state GDP, population and income. Built an R Shiny dashboard with region and year filters, economic-factor comparisons, profit distributions and state rankings.

Source: ETW2001 / Assignment 2

This notebook documents the existing project. Website code cells are presentation controls, not a live Python kernel.

```python
import csv
from collections import defaultdict
from pathlib import Path
rows = list(csv.DictReader((Path.cwd() / "superstore-summary.csv").open()))
sales = defaultdict(float)
for row in rows:
    sales[row["Region"]] += float(row["Sales"])
for region, value in sorted(sales.items(), key=lambda x: -x[1]):
    print(f"{region}: ${value:,.2f}")
```
